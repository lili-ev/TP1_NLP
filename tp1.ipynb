```python
text = "Bonjour ! J'aime beaucoup le NLP, c'est TRÈS intéressant... #Python123"
print(text)
```

```python
text_lower = text.lower()
print(text_lower)
```

```python
import string

text_clean = text_lower.translate(str.maketrans('', '', string.punctuation))
print(text_clean)
```

```python
from nltk.tokenize import word_tokenize

tokens = word_tokenize(text_clean)
print(tokens)
```

```python
from nltk.corpus import stopwords

stop_words = set(stopwords.words('french'))
print(stop_words)
```

```python
tokens_filtrés = [word for word in tokens if word not in stop_words]
print(tokens_filtrés)
```
